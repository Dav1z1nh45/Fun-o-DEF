```python
def calcula_media(notas):
    soma = 0
    for nota in notas:
        soma = soma + nota
    return soma / len(notas)

notas_a = [7, 8, 6]
notas_b = [5, 9, 5]

print(f"{calcula_media(notas_a):.2f}")
print(f"{calcula_media(notas_b):.2f}")
```

```python
def maior_de_idade(idade):
    return idade >= 18

idade = int(input("Digite sua idade: "))

if maior_de_idade(idade):
    print("Acesso permitido")
else:
    print("Acesso negado")
```

```python
#Calculadora de desconto
def calcular(preço, desconto_percentual):
    desconto = preço * (desconto_percentual / 100)
    return preço - desconto

produto = float(input("Digite o valor do produto: "))
desc = float(input("Digite o percentual de desconto: "))

preço_final = calcular(produto, desc)

print(f"Preço final: R${preço_final:.2f}")
```

```python
###Atividade 01

def verificar_aprovacao():
    nota = float(input("Digite a nota: "))
    if nota >= 6:
        return "Aprovado"
    else:
        return "Reprovado"

resultado = verificar_aprovacao()
print(resultado)
```

```python
###Atividade 02
def soma_lista():
    numeros = []

    print("Digite os números (digite 'fim' para parar):")

    while True:
        entrada = input("Número: ")

        if entrada.lower() == "fim":
            break

        numeros.append(float(entrada))

    total = 0
    for numero in numeros:
        total += numero

    return total
    44

resultado = soma_lista()
print("Soma =", resultado)
```

```python
###Atividade 03


def filtrar_maiores_de_idade():
    idades = []

    print("Digite as idades (digite 'fim' para parar):")

    while True:
        entrada = input("Idade: ")

        if entrada.lower() == "fim":
            break

        idades.append(int(entrada))

    maiores = []
    for idade in idades:
        if idade >= 18:
            maiores.append(idade)

    return maiores

resultado = filtrar_maiores_de_idade()
print("Maiores de idade:", resultado)
```

```python

###Atividade 04
def calcular_juros():
    valor = float(input("Valor: "))
    taxa = input("Taxa (Enter = 0.05): ")

    if taxa == "":
        taxa = 0.05
    else:
        taxa = float(taxa)

    return valor * (1 + taxa)

print("Resultado:", calcular_juros())
```

```python
###Atividade 05

def analisar_notas():
    notas = []
    while True:
        n = input("Nota (fim=parar): ")
        if n == "fim": break
        notas.append(float(n))

    return {
        "maior": max(notas),
        "menor": min(notas),
        "media": sum(notas)/len(notas),
        "aprovados": sum(1 for n in notas if n >= 6)
    }

print(analisar_notas())
```
